```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(1, 2, 100)

sinx = np.sin(x)
cosx = np.cos(x)

print(sinx)
print(cosx)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2)
ax1.plot(x, sinx)
ax2.plot(x, cosx)
ax1.set_title('sin(x)')
ax2.set_title('cos(x)')

plt.savefig('CD2Prompt2')
```
