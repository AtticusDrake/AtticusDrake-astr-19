```python
import numpy as np
import math
```

This cell defines and returns sin(x) and cos(x)

```python
x = 1
def sin_x(x):
    return np.sin(x)

def cos_x(x):
    return np.cos(x)
```

This cell computes the values

```python
x_values = np.linspace(0, 2, 1000)

sin_values = sin_x(x_values)
cos_values = cos_x(x_values)
```

This cell creates the table

```python
print(f"{'x':>8} {'sin(x)':>10} {'cos(x)':>10}")
print("-" * 30)

# Loop through and print the first 10 values
for i in range(10):
    print(f"{x_values[i]:8.5f} {sin_values[i]:10.5f} {cos_values[i]:10.5f}")
```
